# Elliptic curves as surfaces in ℝ³

The code lives in [`elliptic_curve_3d.sage`](elliptic_curve_3d.sage) (class `EllipticCurve3D`, see its docstrings); this notebook only uses it. The surface is $(\operatorname{Re} x, \operatorname{Im} x, \operatorname{Re} y)$ for the points $(x, y) = (\wp(z) - b_2/12,\ (\wp'(z) - a_1 x - a_3)/2)$ of the *given* model, over the half fundamental domain $z = s\,\omega_1 + t\,\omega_2$, $-\tfrac12 \le s \le \tfrac12$, $0 \le t \le \tfrac12$.

```python
load("elliptic_curve_3d.sage")
```

```python
S = EllipticCurve3D("20.a3")
G = S.surface(
    colormaps.Greens_r,
    spacing = 1/8,
    cutoff = 0,
    radius = 3,
    opacity = 1/2,
    plot_points=600
)
```

```python
G
```

## Check 1: the surface extends the real points of E

The real points, drawn independently of $\wp$ as $(x, 0, \pm\sqrt{f(x)})$ in the plane $\operatorname{Im} x = 0$, must lie on the surface. (Plotting $(\wp, \wp')$ directly gives the curve $\eta^2 = 4X^3 - g_2 X - g_3$ instead: $E$ shifted by $b_2/12$ in $x$ and stretched by 2 in $y$.)

```python
x = var('x')
E = S.curve()
_, a2, _, a4, a6 = E.ainvs()
f = x^3 + a2*x^2 + a4*x + a6                                   # E:  y^2 = f(x)
e3, e2, e1 = sorted(E.two_division_polynomial().roots(RR, multiplicities=False))

real_points = None
for lo, hi in [(e1, 1.8), (e3, e2)]:
    for s in (1, -1):
        c = parametric_plot3d((x, 0, s*sqrt(abs(f))), (x, lo, hi), color='red', thickness=6)
        real_points = c if real_points is None else real_points + c
G + real_points
```

The same thing from the class: `S.real_points_plot(radius)` draws the rows $t = 0$ and $t = \tfrac12$ of the parametrisation, and `S.plot(...)` combines both.

```python
S.plot(colormaps.Greens_r, radius=3, opacity=1/2, plot_points=400)
```

## Check 2: the same slice, in the plane

Rows $t = 0$ and $t = \tfrac12$ of the grid are the lines on which $x$ is real; they are the two real components of $E$ (red) and must sit on the implicit plot of $y^2 = f(x)$ (blue). See `slice_check.sage` for the full diagnosis, including the original, shifted slice.

```python
x, y = var('x, y')
Sg, Tg, xg, yg = S.grid(400)
P = implicit_plot(y^2 - f, (x, -2.5, 3.5), (y, -6, 6), color='blue', linewidth=2)   # real points of E
for j in (0, -1):                     # rows t = 0 and t = 1/2 of the grid: the slice Im x = 0 of the surface
    xs, ys = xg[:, j].real, yg[:, j].real
    keep = np.abs(xs) < 10
    P += point2d(list(zip(xs[keep][::5], ys[keep][::5])), color='red', size=8)
P.show(xmin=-2.5, xmax=3.5, ymin=-6, ymax=6, figsize=7, axes_labels=['Re x', 'Re y'])
```
